### **Preprocess Dataset Format**

Using the `.head()` command, we observe that the `genres` and `tags` columns are encoded into space separated list, which are not recognised in Python, so we will have to change the format to proper lists.

```python
def colToList (x:str) -> list[str]:
    # Detect the separator
    x = re.sub(r"'[\n ].?'", ",",x)
    
    # Remove the indicator of the raw space separated list and split on coma
    xSplit = x.removeprefix("['").removesuffix("']").split(",")
    
    # Return the list where the text is capitalize (it is just prettier)
    return list(map(str.capitalize, xSplit))

# Apply the split
df_filtered.loc[:,"genres"] = df_filtered.loc[:,"genres"].apply(lambda x: colToList(x))
df_filtered.loc[:,"tags"] = df_filtered.loc[:,"tags"].apply(lambda x: colToList(x))
df_filtered.head()
```

##### **Language description** 
To achieve this goal, we will first examine the most frequent words in our dataset. Then, we will build a function to compare the dataset with and without stop words.

```python
def plotDescriptionLanguage(df:pd.DataFrame, col="description", stopFiltered=True):
    descriptionList = list(map(str,df.loc[:,col].to_list()))
    descriptionString = " ".join(descriptionList)
    # Match all words and digits
    wordDescriptionList = re.findall(r"[\w\d-]+",descriptionString)

    # Remove the stopwords if asked
    wordDescriptionList = [i.capitalize() for i in wordDescriptionList if (i.lower() not in stop_words) or not stopFiltered]

    # Apply the counter
    wordDescriptionListCounter = Counter(wordDescriptionList)

    # Sort the counter
    wordDescriptionListCounterSorted = [[],[]]
    for word in sorted(wordDescriptionListCounter, key=lambda x:wordDescriptionListCounter[x], reverse=True):
        wordDescriptionListCounterSorted[0].append(word)
        wordDescriptionListCounterSorted[1].append(wordDescriptionListCounter[word])

    # Make a plot
    plt.close("all")
    sns.set_style("dark")
    sns.barplot(
        x=wordDescriptionListCounterSorted[1][:40],
        y=wordDescriptionListCounterSorted[0][:40],
        palette="BrBG"    
    )
    plt.title(f"Wordcount {'(stopwords filtered)' if stopFiltered else ''}")
    plt.xlabel("Wordcount")
    plt.ylabel("Words")

    plt.show()
    
    return wordDescriptionListCounterSorted
    
```

```python
# Dictionary of interchangeable genres
transferGenres = {
    "Smut":"Adult",
    "Shoujo ai":"Yuri",
    "Xianxia":"Xuanhuan",
    "Shounen ai":"Yaoi"
}
# Genres to remove
forbiddenGenres = set([
    "Sports",
    "Wuxia",
    "Mecha",
    "Gender bender"
])

# For each document
dfGenresFiltered = []
for genres in df_filtered.loc[:,"genres"]:
    genres_copy:list = genres.copy()
    
    # Changes the genres by removing/transforming the labels
    for genre in genres:
        if genre in forbiddenGenres:
            genres_copy.remove(genre)
        if genre in transferGenres:
            genres_copy.remove(genre)
            genres_copy.append(transferGenres[genre])
    dfGenresFiltered.append(genres_copy)

df_genresFiltered = df_filtered.copy(deep=True)
df_genresFiltered.loc[:, "genres"] = pd.Series(dfGenresFiltered)
```

## 2.2 - Data Split
In this subpart, we discuss data splitting strategies in the context of our multilabel situation. In our case, a classic 60-20-20 split or a stratified split is not necessarily the best option, since multiple labels can appear, making a balanced division difficult.

The initial idea was to create a training, development and test set for each possible label. However, this would combine the training and test data from one classifier to another, and it is impossible to completely avoid interdependence in our context.

This is why we will use [**scikit multilearn**](https://web.archive.org/web/20251114024511/https://scikit.ml/stratification.html) `iterative_train_test_split` function as suggested on the [*Data Science Stack Exchange*](https://datascience.stackexchange.com/questions/45174/how-to-use-sklearn-train-test-split-to-stratify-data-for-multi-label-classificat), which provide an optimised version of the classical train test split for multilabel. Note that, in opposition with standard train-test split method, the `iterative` method, described by ***Konstantinos Sechidis et al.*** in the paper [***On the Stratification of Multi-label Data***](link.springer.com/content/pdf/10.1007/978-3-642-23808-6_10.pdf), is usually deterministic and aims to conserve the balance of the different labels in the produced sets. 

Unfortunately, in the event of a tie, the break is chosen at random. From version 0.2.0 of scikit-multilearn, the implemented function does not enable us to fix the state. As a result, we will temporarily fix the state globally using `np.random.seed` (after checking, this seems to work). For simplicity and due to time constraints, we will evaluate our model using a tie-breaking seed of 0 for the training/dev-test split and a seed of 42 for the dev/test split. Considering the nature of the split method, which is mostly deterministic, this should not have significant consequences.

In our case, with a large number of classes and a reasonably sized dataset, we will use a 75–15–10 repartition.

```python
# To use the function: https://github.com/scikit-multilearn/scikit-multilearn/discussions/282

# Transform our labels lists into dummy classifier
mlb = MultiLabelBinarizer()
Y = mlb.fit_transform(df_genresFiltered.loc[:,"genres"].tolist())

# Transform our X into something working
X = df_genresFiltered[["description"]].to_numpy()
X = X[..., np.newaxis]

# Register the id of each labels (column of y)
labelIdtoLabelName = {i:label for i, label in enumerate(mlb.classes_)}
labelNameToLabelId = {v:k for k,v in labelIdtoLabelName.items()}


# Train Dev Test Split
## Train Test Split (0.75 training)
## Ensure reproducibility globally
np.random.seed(0)
X_train, y_train, X_test, y_test = iterative_train_test_split(X, Y, test_size = 0.25)
## Test Dev split (0.6*0.25=0.15 dev and 0.4*0.25 = 0.1 test)
np.random.seed(42)
X_test, y_test, X_dev, y_dev = iterative_train_test_split(X_test, y_test, test_size=0.6)

# Reset the seed
np.random.seed(None)

# Print some informations:
print(
f"""
The shape of the training set is {X_train.shape},
the shape of the test set is {X_test.shape},
the shape of the dev set is {X_dev.shape}
"""
)
```

## 2.3 - Preprocessing Strategies
In this section, we will present our preprocessing choices in a bullet-point format with justification.
- **Tokenization Method:** Word-level tokenisation. We do not plan to use transformers in our multilabel context as this approach would be too time-consuming without a GPU. Therefore, we will stick to classical machine learning (ML) techniques such as TF-IDF, Bayes and logistic regressions, as well as pre-trained embeddings like GloVe, which is not trained on subwords. Additionally, in Section 2.1 bis, we observed many composed words with rare occurrences. Consequently, we will remove the hyphen between composed words to break them down into separate tokens. Also, we will lowercase the words to standardise different versions of the same word.
- **Lemmatization vs Stemming:** In our case, where we hypothesise that the context of a label is present in the semantics of the tokens, and where we can handle higher time complexity because the texts are not too long, we have chosen to use lemmatisation. Indeed, it could yield better results than stemming because it is context-aware, unlike stemming.
- **Stopwords strategy:** In Section 2.1bis, we observed that classical stop words (such as *A*, *The*, *To*, *And*, etc.) are very frequent without conveying a particular meaning. In this case, we will use the nltk English stopword set to filter them out.
- **Rare word strategy:** In Section 2.1bis, we discovered that more than half of the words are very rare and/or specialised. In this context, we could use TF-IDF to identify and filter out the less important words based on a proportion or threshold.
- **Handling of emoji and punctuation:** Since we are not conducting a sentiment analysis, emojis (which are almost non-existent in our dataset) and punctuation do not reveal particularly strong semantics. We will therefore remove them. The same will be done for numbers in general.

With that in mind, we can now preprocess our dataset.

```python
def preprocess(string:str, morphologicalAnalysis:bool = False) -> np.ndarray[str]:
    # Preprocess the string using SpaCy and en_cor_web_sm
    string = nlp(str(string))
    
    # If morphological analysis, stop here
    if morphologicalAnalysis:
        return string
    
    # Get the lemma from the 
    lemmatisedString = " ".join([token.lemma_ for token in string])
    
    # Match all words that are not digits. Match accentuated characters and words with apostrophes (don't, won't ,...)
    strings = re.findall(r"[^\W\d_]+(?:'[^\W\d_]+)*", lemmatisedString)
    
    # Return and remove stop words
    return [word.lower() for word in strings if word.lower() not in stop_words]

# Just so that we know the time spent for preprocessing
begin = time.time()

# Preprocess the test set
preprocessedX_test = [preprocess(X_test[i,0,0]) for i in range(X_test.shape[0])]
print(f"Preprocessing of Test Set done! [{X_test.shape[0]} in {time.time()-begin:.4f}s]")
begin = time.time()

# Preprocess the dev set
preprocessedX_dev = [preprocess(X_dev[i,0,0]) for i in range(X_dev.shape[0])]
print(f"Preprocessing of Dev Set done! [{X_dev.shape[0]} in {time.time()-begin:.4f}s]")
begin = time.time()

# Preprocess the training set
preprocessedX_train = [preprocess(X_train[i,0,0]) for i in range(X_train.shape[0])]
print(f"Preprocessing of Train Set done! [{X_train.shape[0]} in {time.time()-begin:.4f}s]")
```

Now that we have successfully preprocessed the text, the next step is to tokenise it. To do this, we will first create a vocabulary based on the training set.

```python
# Counter of word from training set
trainingWordCounter = Counter(chain.from_iterable(preprocessedX_train))

# Get a vectorizer
vectorizer = TfidfVectorizer(
    tokenizer = lambda x: x, # Input as tokens
    preprocessor = lambda x: x, # Skip preprocessing (already done)
    token_pattern = None # Disable tokenizer (we already have tokens) 
)

# Get TfIdf
preprocessingTFIDFMatrix = vectorizer.fit_transform(preprocessedX_train)

# Get the vocabulary mapping
trainingVocab = vectorizer.vocabulary_
inversedTrainingVocab = {v:k for k,v in trainingVocab.items()}

# Get the initial size of the training vocabulary
print(f"Size of the vocabulary: {len(trainingVocab)}")
```

From now on, we have the base vocabulary. However, it is still too large. Even after removing stop words, lemmatising and selecting only the training set vocabulary, we went from 41,025 tokens in the whole dataset to 26,449.

To solve this issue, we will set a threshold for the average TF-IDF per token. Indeed, less important tokens tend to have lower tf-idf values.

```python
# The minimal threshold on tf-idf for which a token is not considered rare anymore. 
# This threshold is purely arbitrary and could be used as an hyperparameter if the time/Processing time is on our side
# This threshold was chose to observe token that are not atrociously rare while being important in the semantics. By this
# threshold, we have approximately 2.2k vocabulary size, which is also reasonable for our system. 
threshold = 0.0005 #0.0005

# Get the vocab above the threshold in term of average tf-idf
filteredVocabID = np.argwhere(preprocessingTFIDFMatrix.mean(axis=0)>threshold)[:,1]
filterVocab = set([inversedTrainingVocab[i] for i in filteredVocabID])

# Plot the results of the selection
pd.DataFrame(
    list(sorted([(
        vocab, 
        trainingWordCounter[vocab],
        preprocessingTFIDFMatrix[:, trainingVocab[vocab]].mean(axis=0).round(6).tolist()[0][0]
    ) for vocab in filterVocab], key=lambda x: x[2], reverse=True)),
    columns=["Token", "Frequency", "Average tf-idf"]
)
```

GloVe's particular feature is that its embedding can understand the underlying linear structures between words. For example, 'King-Man' plus 'Woman' would be approximately equal to 'Queen'. This structure will be useful for providing a document embedding, which we will use for classification.

Note (How to download GloVe): when downloaded, unzip it, and place the obtained file in the folder `GloVe` after renaming it into `GloVe2024.txt` 

```python
# Load GloVe
# Solution adapted from https://stackoverflow.com/questions/37793118/load-pretrained-glove-vectors-in-python
class GloVeClass():
    def __init__(self, path: str):
        self.GloVe = pd.read_table(path, sep=" ", index_col=0, header=None, quoting=csv.QUOTE_NONE)
        self.GloVeMatrix = self.GloVe.to_numpy()
        self.dim = self.GloVe.shape[1]

        # Precompute normalized matrix to make closest word faster
        norms = np.linalg.norm(self.GloVeMatrix, axis=1, keepdims=True)
        self.GloVeMatrixNorm = self.GloVeMatrix / norms

    def closestWord(self, vec: np.ndarray) -> str:
        # Normalise the input vector
        vec_norm = vec / np.linalg.norm(vec)
        
        # Compute cosine similarity as only dot product since it is already normalised (||A||*||B||) 
        cos = np.dot(self.GloVeMatrixNorm, vec_norm)
        
        # Get the index with the highest cosine similarity
        i = np.argmax(cos)
        
        # Get the word
        return self.GloVe.iloc[i].name

    def getVector(self, tokens:list[str]) -> np.ndarray:
        # If the list is empty, return nan for the document embedding
        if len(tokens) == 0:
            return np.full((1, self.dim), np.nan)
        
        # By defaults, embedding is unknown
        result = np.full((len(tokens), self.dim), np.nan)

        # Find which tokens exist
        mask = [t in self.GloVe.index for t in tokens]

        # If tokens exists
        if any(mask):
            # Get the terms
            found_tokens = [t for t, m in zip(tokens, mask) if m]
            
            # Place the vector at the right position
            result[mask] = self.GloVe.loc[found_tokens].to_numpy()

        return result
        
    def getCosineSimilarity(self, vec1:np.ndarray, vec2:np.ndarray) -> float:
        return np.dot(vec1, vec2)/(np.linalg.norm(vec1)*np.linalg.norm(vec2))
        
# Check whether GloVe was well imported
GloVe = GloVeClass(path="GloVe/GloVe2024.txt")
```

To make classifications in cases involving multilabel data, we will use scikit-multilearn to wrap scikit-model execution. In short, rather than training one classifier per column ourselves, we will let skmultilearn do it to optimise the process.

As we can see here, the Bayes classifier worked with our TF-IDF encoding, which can be interpreted as a count in some way. Unfortunately, this is not the case for GloVe, which possesses negative values in document embeddings that cannot therefore be interpreted as a count. This causes the classifier to fail for this representation. We will therefore use a different baseline: the closeness to the average embedding (Nearest neighbour in a sense).

```python
class ClosenessDocEmbeddingClassifier():
    def __init__(self):
        pass
    
    def fit(self, X_train:np.ndarray[float], Y_train:np.ndarray[int]) -> None:
        # Initialise the label embedding
        self.averageEmbedding = {}
        
        # For each label
        for label in range(Y_train.shape[1]):
            # Find which rows has this label
            inLabel = np.where(Y_train[:,label])[0]
            
            # Take the average of document possessing it
            self.averageEmbedding[label] = np.mean(X_train[inLabel,:], axis=0)
    
    def calibrateThreshold(self, X_dev:np.ndarray[float], Y_dev:np.ndarray[int], precision:int=10**4, verbose:bool = True) -> None:
        # Cut the space [0,1] to precision slice (to test all the thresholds)
        thresholds = np.linspace(0,1,precision)
        
        # Register the best Hamming similarity
        bestHamming = np.full(len(self.averageEmbedding.keys()), -1, dtype=float)
        self.thresholds = np.full(len(self.averageEmbedding.keys()), -1, dtype=float)
        begin = time.time()
        
        # For each threshold
        for i, threshold in enumerate(thresholds):
            # Warn the user if necessary
            if verbose and i%1000 == 0:
                print(f"[{time.time()-begin:.5f}s] threshold: {i}/{precision}")
            
            # Create the predicted matrix as 0
            Y_pred = np.zeros_like(Y_dev, dtype=int)
            
            # For each label
            for label, labelEmbedding in self.averageEmbedding.items():
                # Compute the cosine similarity
                labelCos = GloVe.getCosineSimilarity(X_dev,labelEmbedding)
                
                # Fill the prediction
                Y_pred[np.where(labelCos>=threshold)[0], label] = 1
                
                # Take Hamming similarity for each label and store it if it is the best
                if 1 - hamming_loss(Y_dev[:,label], Y_pred[:,label]) > bestHamming[label]:
                    self.thresholds[label] = threshold
    
    def predict(self, X_test:np.ndarray[float]) -> np.ndarray[int]:
        # Initialise the predicted matrix as zeros
        Y_pred = np.zeros((X_test.shape[0], len(self.averageEmbedding.keys())), dtype=int)
        
        # For each labels
        for label, labelEmbedding in self.averageEmbedding.items():
            
            # Compute cosine similarity
            labelCos = GloVe.getCosineSimilarity(X_test, labelEmbedding)
            
            # Make predictions 
            Y_pred[np.where(labelCos>=self.thresholds[label])[0], label] = 1
        
        # Return the prediction
        return Y_pred
            
```

Now, we can make our GloVe predictions. Please note that hyperparameter tuning is performed using the Hamming similarity due to the imbalance in the labels and the greater importance assigned to the more frequent ones. 

**Warning: since this can be long and/or memory intensive, you can go to the next restoration point for the results.**

```python
# Create the classifier
embeddingClassifier = ClosenessDocEmbeddingClassifier()

# Fit the classifier and tune hyperparameter 
embeddingClassifier.fit(X_train_GloVeVectorisedOutNA, y_trainGloVeVectorisedOutNA)
embeddingClassifier.calibrateThreshold(X_dev_GloVeVectorised, y_dev)

# Predict the results
y_predEmbeddingGloVe = embeddingClassifier.predict(X_test_GloVeVectorised)
```

#### **Report**
Now that we have computed all the baselines, we should evaluate them. Typically, we present the confusion matrix alongside the accuracy, precision, recall and F1 score. Unfortunately, this is not feasible in the context of multilabel classification. Indeed, a confusion matrix with all crossings would have 2**27 rows, which is too large, and an aggregated confusion matrix with all labels would be too misleading.

In terms of accuracy, we cannot say that a combination of genres is incorrect just because one of the seven is missing, nor can we say that one has been predicted despite not being present. However, we can compare the accuracy of the predictions for each label. Ultimately, we will conduct this analysis:

- Per label: Accuracy, F1-Score, Precision, Recall
- Per rows: Jaccard similarity, Hamming Similarity, label accuracy product.

Here, we will not reexplain what accuracy, F1-Score, Precision and Recall are, but we will motivate the usages of the aggregation we will use.

First, the Jaccard similarity is computed as $J(Y,\hat{Y})=\frac{\#Y\cap\hat{Y}}{\#Y\cup\hat{Y}}$, where $\hat{Y}$ is the predicted genres and $Y$ the real genres. This metric is valued to 1 if the sets are identical, 0 if they are disjoint. It is convenient as it penalises both adding unnecessary genres, but also the absence of some. To get a value for the whole test set, we will average the row Jaccard similarity.

Second, the Hamming "Similarity" that we will define as: $H(Y,\hat{Y}) = 1 - \frac{H(Y,\hat{Y})}{\#Y}$, where $H$ is the Hamming distance measuring the number of differences. This metric is valued at 1 if there is no difference between the vectors and at 0 if all is different. The difference with the Jaccard similarity is that it is sensitive to the number of 0, i.e. the sparser the vector, the more biased the result. As for Jaccard, we will take the average.

Finally, we will proceed with the product of accuracies per label. In a sense, if I can predict with an accuracy of 0.9 the genre `Action` and 0.7 the genre `Fantasy`, and ..., then by making the product, we would take into account all the errors. Unfortunately, it could be less meaningful compared to the other metrics, as it will quickly collapse to 0 considering the number of labels to predict (example: 90% of accuracy over all labels -> $0.9^{27}\approx0.0581$).

Let first proceed with the classic metric per labels.

```python
# Function to compute the metrics
def computeLabelMetrics(y_true:np.ndarray[int], y_pred:np.ndarray[int], classes_:list[str]) -> pd.DataFrame:
    # Initialise the dataframe
    genreMetrics = pd.DataFrame(columns=["Genre", "Accuracy", "Precision", "Recall", "F1-Score"])
    
    # If it is a sparse matrix, convert it back into array
    if not isinstance(y_pred, np.ndarray):
        y_pred = y_pred.toarray()
    
    # For each label
    for label in range(y_true.shape[1]):
        # Compute the metrics
        labelMetric = [
            classes_[label],
            accuracy_score(y_true[:,label], y_pred[:,label]),
            precision_score(y_true[:,label], y_pred[:,label], zero_division=np.nan),
            recall_score(y_true[:,label], y_pred[:,label]),
            f1_score(y_true[:,label], y_pred[:,label])
        ]
        # Register it
        genreMetrics.loc[label] = labelMetric
    
    # Return the dataframe
    return genreMetrics

# Compute the metrics for TFIDF over the 3 classifiers
labelMetricsBayesTfidf = computeLabelMetrics(y_test, y_predBayesTfidf, classes_)
labelMetricsLogisticTfidf = computeLabelMetrics(y_test, y_predLogisticTfidf, classes_)
labelMetricsEmbeddingTfidf = computeLabelMetrics(y_test, y_predEmbeddingTfidf, classes_)

# Compute the metrics for GloVe over the two classifiers
labelMetricsLogisticGloVe = computeLabelMetrics(y_test, y_predLogisticGloVe, classes_)
labelMetricsEmbeddingGloVe = computeLabelMetrics(y_test, y_predEmbeddingGloVe, classes_)
```

Now let's compute our row metrics.

```python
# Function to compute the metrics
def computeRowMetrics(X_unprocessed:np.ndarray[str], y_true:np.ndarray[int], y_pred:np.ndarray[int], classes_:list[str]) -> pd.DataFrame:
    # Initialise the dataframe
    observationMetrics = pd.DataFrame(columns=["Description", "Genres", "Predicted Genres", "Jaccard", "Hamming"])
    classes_ = np.array(classes_)
    
    # If it is a sparse matrix, convert it back into array
    if not isinstance(y_pred, np.ndarray):
        y_pred = y_pred.toarray()
    
    # Compute the metric for each observations
    for i, description in enumerate(X_unprocessed):
        rowMetric = [
            description,
            list(classes_[np.where(y_true[i,:])[0]]),
            list(classes_[np.where(y_pred[i,:])[0]]),
            jaccard_score(y_true[i,:], y_pred[i,:]),
            1 - hamming_loss(y_true[i,:], y_pred[i,:])
        ]
    
        # Register it
        observationMetrics.loc[i] = rowMetric
    
    # Return the dataframe
    return observationMetrics.sort_values("Jaccard", ascending = False)
```

This seems to work, we have 2 document embedding of size 768. Now let us transform our text into BERT embeddings.

```python
def getBertEmbedding(Xset:np.ndarray[str], batchSize:int = 25) -> list[list[float]]:

    # Initialise the list that will contains the embeddings
    Xset_embedded = []
    begin = time.time()

    # Put the Bert Model in evaluation mode
    modelBert.eval()
    # For each batch
    for i in range(0, (Xset.shape[0]/batchSize).__ceil__()):
        # Tokenize the batch
        XsetBertTokenized = tokenizerBert(
            # Get str to always have a valid sequence
            list(map(str, Xset[i*batchSize:(i+1)*batchSize,0])),
            return_tensors="pt",
            padding=True, 
            truncation=True
        )
        # Print the advancement
        if i%10 == 0:
            print(f"Batch no. {i}/{(Xset.shape[0]/batchSize).__ceil__()} [{time.time()-begin:.4f}s]")
        
        # Get the embedding
        with torch.no_grad():
            Xset_embedded.extend(modelBert(**{k:v.to("cuda") for k,v in XsetBertTokenized.items()}).last_hidden_state[:,0,:].tolist())
        
    return Xset_embedded
```

Now that everything seems to be working, we can proceed with neural networks. Please note that multiple undisplayed iterations were made before achieving the following network. Here is the neural network that we will use:

```python
class NeuralMultiLabelClassifier(nn.Module):
    # 768: dimension of the BERT embedding, 27: dimension of the label space
    def __init__(self, input_dim:int=768, num_labels:int=27, refine:bool = False):
        super().__init__()
        
        # Main component to identify the features (input)
        self.feature = nn.Sequential(
            nn.Linear(input_dim, 512),
            nn.LayerNorm(512),
            nn.GELU(),
            nn.Dropout(0.3)
        )
        
        # Output layer
        self.classifier = nn.Linear(512, num_labels)
        
        # Linear Correlation learning module + the weight we will derive in the logits
        self.label_correlation = nn.Sequential(
            nn.Linear(num_labels, num_labels, bias=False),
            nn.Tanh()
        )
        self.corr_gate = nn.Parameter(torch.tensor(0.5))
        
        # Refining layer (is active only if correlation is not). Used to capture higher correlation relationships between labels 
        self.addRefine = refine
        self.refine = nn.Sequential(
            nn.Linear(512 + num_labels, 256),
            nn.GELU(),
            nn.Linear(256, num_labels)
        )

    # Feed forward function
    def forward(self, x):
        # Get the feetures
        h = self.feature(x)
        # Classify
        logits = self.classifier(h)
        
        # Add the correlation between the labels
        if not self.addRefine:
            logits = logits + self.corr_gate*self.label_correlation(logits)
        else:
            logits = logits + self.refine(torch.cat([h, logits], dim=1))
        return logits
```

Now, we must explain and justify our architectural design. The architecture itself is divided into three consecutive parts.
- **The Feature layer:** Constitute the hidden layer in order to learn the features associated with each token. It makes use of three main parts: a dense layer to learn the components; a normalisation layer to stabilise the model and ensure convergence; and a GELU (Gaussian Error Linear Unit) activation function, which apparently improves performance in Transformer contexts, as suggested [here](https://www.baeldung.com/cs/gelu-activation-function) (BERT uses a Transformer, so it makes sense to use this activation function here). It also uses a 30% dropout layer to avoid overfitting (this dropout is used because our model is relatively simple, and the standard range of values is between 20% and 50%, as seen [here](https://medium.com/@piyushkashyap045/understanding-dropout-in-deep-learning-a-guide-to-reducing-overfitting-26cbb68d5575)). In a sense, it could be a tuneable parameter, but this will not be exploited here.
- **The Classifier layer:** This layer connects the networks to the 27 output labels. We do not perform softmax here, as this would eliminate the ability to predict multiple labels simultaneously. In order to interpret the results, we must first find the optimal threshold at which a genre is considered to have been acquired.
- **(Choice 1) The Correlation Layer:** Passing through the first two layers ensured that we could use the correlation between tokens. However, this does not take into account the correlation between labels. This is why we apply this layer: to learn the correlation between the label and apply corrections to the results. Note that we use Tanh afterwards for the same reason of normalisation (here, between -1 and 1 to respect the meaning of the correlation); also, the potential noise created by this layer is mitigated using a learnable parameter. Basically, this layer says: "If I increase A, then I should also increase/decrease B".
- **(Choice 2) The Refinement Layer:** By using a refinement layer instead of a correlation layer, we are not limited to direct correlation; we can also investigate nonlinear, deeper patterns, such as conditional or multi-correlated patterns. Therefore, we use a dense layer with dimensions of 512+27 to take the features and labels into account.

In the end, we have an architecture like this:
1. **Raw documents**
2. **Bert Embeddings**
3. **Feature Layer**
4. **Classifier Layer**
5. **Correlation | Refinement Layer**
6. **Sigmoid (prediction)**
7. **Thresholds (to go back to binary)**

For the tuneable parameters, we have the dropout rate, which we will not be adjusted, and, more importantly, the thresholds. Each output may have a different distribution. Consequently, we cannot set a global threshold such as *IF pred>0.5 THEN PREDICT 1*; this would not make sense and would not take into account the label imbalance, which is probably the biggest challenge here, as we will explain.

Now that we explained the architecture, let us normalise our embeddings:

```python
# Normalise the Bert Embedding
X_train_BertEmbeddedNormalised = nn.functional.normalize(torch.tensor(X_train_BertEmbedded))
X_dev_BertEmbeddedNormalised = nn.functional.normalize(torch.tensor(X_dev_BertEmbedded))
X_test_BertEmbeddedNormalised = nn.functional.normalize(torch.tensor(X_test_BertEmbedded))
```

Now, we proceed with the training loop. Here, we will do the training for both model with correlation layer and refinement layer.

```python
def trainNN(X_train:torch.tensor, Y_train:torch.tensor, batchSize:int = 100, epochs:int=50, refine:bool=True, seed:int=0) -> NeuralMultiLabelClassifier:
    # Setup the seed of torch for reproducibility
    torch.manual_seed(seed)
    
    # Because the labels are unbalanced, use the Weighted BCEWithLogitLoss so that we can add a higher weight to rare labels
    positiveCount = Y_train.sum(axis=0)
    negativeCount = Y_train.shape[0] - positiveCount
    positiveWeight = negativeCount / positiveCount
    criterion = nn.BCEWithLogitsLoss(pos_weight=torch.tensor(positiveWeight))
    
    # Initialise the Neural Model
    neuralModel = NeuralMultiLabelClassifier(refine=refine)
    
    # Use classic Adam with classic learning rate
    optimizer = torch.optim.Adam(neuralModel.parameters(), lr=1e-4)

    # Start the training loop
    begin = time.time()
    neuralModel.train()
    for epoch in range(epochs):
        # Inform of the advancement
        if epoch%10 == 0:
            print(f"[{time.time()-begin:.3f}s] {epoch=}")
        
        # Permute the batches at each epoch to avoid Overfitting. The random seed is set to epoch to keep determinism
        rng = np.random.RandomState(epoch)
        permutation = rng.permutation(np.arange(X_train.shape[0]))
        X_epoch = X_train[permutation]
        Y_epoch = Y_train[permutation]
        
        # For each batch
        for batch in range(0, (X_train.shape[0]/batchSize).__ceil__()):
            # Reinitialise gradient
            optimizer.zero_grad()

            # Compute logits and loss for the current batch
            logits = neuralModel(torch.tensor(X_epoch[batch*batchSize:(batch+1)*batchSize,:]))
            loss = criterion(logits, torch.tensor(Y_epoch[batch*batchSize:(batch+1)*batchSize]).float())

            # Back propagation
            loss.backward()
            optimizer.step()
    
    # Return the model
    return neuralModel

# Get a model without refinement
print("[NORMAL TRAINING]")
neuralModelBaseline = trainNN(X_train_BertEmbeddedNormalised, y_train, refine=False)

# Get a model with refinement
print("[REFINE TRAINING]")
refinedNeuralModel = trainNN(X_train_BertEmbeddedNormalised, y_train, refine=True)
```

As mentioned earlier, there are two main challenges in this classification task: the high correlation between the labels, which has already been addressed in the architecture, and the imbalance in the labels, which was observed at the start of the project. 

To address the latter, we use a specific loss function: Weighted Binary Cross Entropy with Logits, as suggested [here](https://stackoverflow.com/questions/61879612/how-to-handle-class-imbalance-in-multi-label-classification-using-pytorch). We use an inverse weight on positive labels to prioritise rare labels over frequent ones (to avoid making the same mistake as with the GloVe embedding earlier, where rare labels were never predicted). 

Regarding the optimiser, we opted for the standard Adam with a standard learning rate.

To avoid overfitting, we only trained the model for 50 epochs, which seems appropriate for a simple model like this one, and permuted the order of the batches at every new epoch to prevent the model from learning an order or pattern within the batches. The seed of the permutation is set to the epoch value to maintain reproducibility.

Now that this is complete, we can discuss one of the most important aspects: threshold tuning.

```python
def tuneOutputThresholds(y_true:np.ndarray[int], y_pred:np.ndarray[int], classes_:list[str], precision:float=10**4) -> np.ndarray[float]:
    thresholds = []
    begin = time.time()

    # For each label
    for label in range(y_true.shape[1]):
        # Initialise the threshold and MacroF1 (we use macroF1 because of class imbalances)
        bestThreshold, bestF1 = 0.5, 0

        # For each threshold
        for threshold in np.linspace(0, 1, precision):
            # Make the prediction based on the threshold
            preds = (y_pred[:, label] > threshold).astype(int)

            # Compute the MacroF1 for the given label
            f1 = f1_score(y_true[:,label], preds, average="macro")

            # If it is the best F1-Score, update the threshold
            if f1 > bestF1:
                bestF1 = f1
                bestThreshold = threshold

        # Register the final threshold
        thresholds.append(bestThreshold)
        
        # Inform the user
        print(f"[{time.time()-begin:.4f}s] Label {label+1}/{y_true.shape[1]} ({classes_[label]}) done. Best threshold: {bestThreshold:.6f}")

    # Return the thresholds as numpy arrays
    return np.array(thresholds)
```

For threshold tuning, we simply find the thresholds (with 0.0001 precision) that maximise the F1 score for the labels in question. Each label has its own threshold to match its distribution.

Note that we do not use accuracy for tuning. Indeed, when examining GloVe, we already saw that accuracy was not a good metric. From this observation, we use the F1 score, which is more resistant to class imbalance. Initially, we tried to maximise the Jaccard index, but this proved counterproductive as it penalised too many false positives for rare labels, leading to no predictions for rare labels.

In this work, we explored the ***Web/Light Novel Dataset*** and used various tools, including Bayes, Logistic, GloVe and BERT-based methods, to predict the genre of a web/light novel from its description. Concerning **area of improvements**, we clearly see four possible ways:
1. Binary relevance over Bayes or logistic regression does not take correlation into account. It would be interesting to see if inferring fuzzy association rules between labels could solve this issue. This would mean that labels would no longer be either fully true or false, but could take values from 0 (fully false) to 1 (fully true). While this hypothesis is interesting because it is closer to reality, we would also need to find a fuzzification technique (probably rule-based) to achieve this, which could be quite challenging.

2. In this report, our focus was on genres, but the dataset actually contains tags, which are essentially subgenres. It could therefore be interesting to see if leveraging these tags could help us infer the labels in some way. However, this could be challenging given the 891 additional subgenres, each with their own imbalances and correlations.

3. Similarly, we could see if adding the novel's title to the classification problem would make a difference, particularly in the WN/LN industry where titles that describe the plot, such as "*I Was Reincarnated as the 7th Prince so I Can Take My Time Perfecting My Magical Ability*" or "*Is It Wrong to Try to Pick Up Girls in a Dungeon?*", are very common.

4. We observed a lot of confusion between the tags. In this respect, we could try merging genres into higher families more thoroughly, or take this confusion into account in the evaluation. For example, we could say: "If I infer `Fantasy` instead of `Supernatural`, it is as if I had inferred `Supernatural` with 70% certainty." (Fuzzy evaluation). This would be an interesting approach and would probably be the best way to represent non-semantically mergeable overlapping genres (e.g. `Adult`!=`Romance`).
